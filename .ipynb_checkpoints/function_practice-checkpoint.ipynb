```python
def even_odd(n):
    if n%2==0:
        return ('even')
    else:
        return ('odd')
even_odd(22)
```

```python
import random
def even_odd(num):
    print(num)
    if num%2==0:
        return ('even')
    else:
        return ('odd')
num=random.randint(1,50)
even_odd(num)
```

```python
def maximum(a,b):
    if a>b:
        return a
    else:
        return b
maximum(34,100)
```

```python
# simple interest
def interest(p,t=2,r=11):
    return{(p*t*r)/100}

interest(10000)
```

```python
def average(a, b, c):
    return round((a + b + c) / 3, 2)

average(10, 20, 25)
```

```python
def leap_year(n):
    if n%4==0:
        return "leap year"
    else:
        return "not a leap year"

leap_year(2024)
```

```python
def year(n):
    if n%400==0 or (n%4==0 and n%100!=0):
        return "leap year"
    else:
        return "not a leap year"

year(2000)
```

```python
# swapping
def swap(a,b):
    temp=a
    a=b
    b=temp
    return a,b

a=eval(input('enter the a:'))
b=eval(input('enter the b:'))
a,b=swap(a,b)
print("a =", a)
print("b =", b)
```
